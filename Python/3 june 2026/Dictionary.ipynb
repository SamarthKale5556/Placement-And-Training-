```python
translate={}                   # translate english words to german using def functions

translate["good"]="gut"
translate["thanks"]="danke"
def eng2german(word):
    if word not in translate:
        return "word not found"
    return translate[word]

print(eng2german("thanks"))
```

```python
emojis={
    "happy":"😊",
    "sad":"😔",
    "Love":"😘"}

def translate(word):
    word=word.lower()
    if word in emojis:
        return emojis[word]
    else:
        return "emoji not found"


print(translate("sad"))
```

```python
import matplotlib.pyplot as plt

text = input("Enter any word or sentence: ")

letters = []
counts = []

for ch in text:
    if ch != " ":   # ignore spaces
        letters.append(ch)
        counts.append(text.count(ch))

print("\nCharacter Counts:")
for i in range(len(letters)):
    print(letters[i], "=", counts[i])

plt.bar(letters, counts)

plt.xlabel("Characters")
plt.ylabel("Count")
plt.title("Character Frequency Graph")

plt.show()
```
