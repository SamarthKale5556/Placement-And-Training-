```python
def average(a,b):       # average of two numbers
    return (a+b)/2
print(average(3,4))
```

```python
def average(*x):       # add *x when multiple arguments is there and you want to continue them ,no need to define parameters in function see above example
    return x
print(average(1,2))
```

```python
def average(*x):     # addition of two numbers using sum function      
    return sum(x)
print(average(1,2))
```

```python
def average(*x):            # to find average of N numbers      
    return sum(x)/len(x)
print(average(1,2))
```

```python
def count_special(s):           # count special characters
    count=0
    for i in s:
        if not i.isalnum() or i.isspace():         # if not numbers and not space then calculate special characters
            count+=1
    return count 

print(count_special("3gagss!2$tuy@%"))
```
